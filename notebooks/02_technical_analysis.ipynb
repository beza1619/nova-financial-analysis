```python
# Task 2: Technical Indicators Analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Technical Analysis Started")
```

```python
# Calculate Technical Indicators
def calculate_technical_indicators(df):
    # Moving Averages
    df['MA_20'] = df['Close'].rolling(window=20).mean()
    df['MA_50'] = df['Close'].rolling(window=50).mean()
    
    # RSI
    delta = df['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI'] = 100 - (100 / (1 + rs))
    
    print("✅ Technical indicators calculated: MA, RSI")
    return df

# Sample data
dates = pd.date_range('2024-01-01', periods=100)
stock_data = pd.DataFrame({
    'Date': dates,
    'Close': 100 + np.cumsum(np.random.randn(100) * 2)
})

stock_data = calculate_technical_indicators(stock_data)
print(stock_data.tail())
```

```python
# Task 2: Technical Indicators Analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("📊 Task 2: Technical Indicators Analysis")

# Fix path issue - go up one level if running from notebooks folder
current_dir = os.getcwd()
print(f"Current directory: {current_dir}")

# Check if we're in notebooks folder and need to go up
if current_dir.endswith('notebooks'):
    data_dir = '../data'
else:
    data_dir = 'data'

print(f"Looking for data in: {data_dir}")

if os.path.exists(data_dir):
    files = os.listdir(data_dir)
    print("✅ Data folder found! Files:", files)
else:
    print("❌ Data folder not found at expected location")
    files = []

# Use sample data for demonstration
print("📝 Using sample data for technical indicators demonstration")

# Create sample stock data
dates = pd.date_range('2024-01-01', periods=100)
sample_data = pd.DataFrame({
    'Date': dates,
    'Close': 100 + np.cumsum(np.random.randn(100) * 2),
    'Volume': np.random.randint(1000000, 5000000, 100)
})

# Calculate Technical Indicators
def calculate_technical_indicators(df):
    # Moving Averages
    df['MA_20'] = df['Close'].rolling(window=20).mean()
    df['MA_50'] = df['Close'].rolling(window=50).mean()
    
    # RSI
    delta = df['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI'] = 100 - (100 / (1 + rs))
    
    # MACD
    exp1 = df['Close'].ewm(span=12).mean()
    exp2 = df['Close'].ewm(span=26).mean()
    df['MACD'] = exp1 - exp2
    df['MACD_Signal'] = df['MACD'].ewm(span=9).mean()
    
    print("✅ Technical indicators calculated: MA_20, MA_50, RSI, MACD")
    return df

# Apply to sample data
technical_data = calculate_technical_indicators(sample_data)

print("\n📈 Sample of Technical Indicators:")
print(technical_data[['Date', 'Close', 'MA_20', 'RSI', 'MACD']].tail())

print("\n✅ Task 2 Completed: Technical indicators ready for real data")
```
